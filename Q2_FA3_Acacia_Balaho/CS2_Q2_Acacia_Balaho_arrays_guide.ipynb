## 📦 Part 1: What is an Array / List?

Imagine you are managing an online game inventory, a class record, or a music playlist.
- **Without Arrays:** You would need separate variables like `item1 = "Notebook"`, `item2 = "Pen"`, `item3 = "Calculator"`, etc. This gets messy very quickly!
- **With Arrays / Lists:** You can group all related items into **one single variable**:
  ```python
  backpack = ["Notebook", "Pen", "Calculator", "Laptop", "Book"]
  ```

### 💡 Real-World Analogy: A Row of Boxes
Think of a list as a row of numbered boxes sitting on a shelf. Each box holds **one item**, and each box has a position number called an **index**.

Let's run the code cell below to create our first Python lists!

````python
# Creating our first Python Lists (Datasets from Study Guide)

# Dataset A: School Backpack Items (Strings)
backpack = ["Notebook", "Pen", "Calculator", "Laptop", "Book"]

# Dataset B: Class Quiz Scores (Numbers)
quiz_scores = [85, 90, 78, 92, 88]

# Dataset C: Favorite Foods (Strings)
favorite_foods = ["Pizza", "Burger", "Pasta", "Fried Chicken", "Ice Cream"]

print("Backpack Items:", backpack)
print("Quiz Scores:", quiz_scores)
print("Favorite Foods:", favorite_foods)
````

## 🔢 Part 2: Understanding 0-Based Indexing

In Python (and most programming languages like C++, Java, and JavaScript), **indexes always start counting at 0**, NOT 1!

| Index Position | `backpack` Item | `quiz_scores` Value |
| :---: | :---: | :---: |
| **`0`** | `"Notebook"` | `85` |
| **`1`** | `"Pen"` | `90` |
| **`2`** | `"Calculator"` | `78` |
| **`3`** | `"Laptop"` | `92` |
| **`4`** | `"Book"` | `88` |

### 🛠️ Accessing and Modifying Items with Square Brackets `[ ]`
- **To access an item:** `list_name[index]`
- **To modify an item:** `list_name[index] = new_value`

Run the code below to see direct indexing in action!

````python
# Activity 1: Direct Index Accessing

#Create your list here
Desktop_SetUp = ['Chrome (or any browser)', 'Roblox', 'Roblox Studios', 'Steam', 'Discord']

#access your list here
print("--- Direct Index Access ---")
print("Element at index 0:", Desktop_SetUp[0])
print("Element at index 1:", Desktop_SetUp[1])
print("Element at index 2:", Desktop_SetUp[2])
print("Element at index 3:", Desktop_SetUp[3])
print("Element at index 4:", Desktop_SetUp[4])

print("\n--- Modifying an Element via Index ---")
print("Original index 2 item:", Desktop_SetUp[2])

# Replacing index 2 "Roblox Studios" with new Value
Desktop_SetUp[2] = "VS Code"

print("Updated index 2 item:", Desktop_SetUp[2])
print("Updated  List:", Desktop_SetUp)
````

## 🔄 Part 3: Array Traversal Using Loops

While direct indexing is great for single items, **traversal** means visiting every element in an array one by one.

We use `for` loops to make traversal fast, clean, and efficient!

### Two Ways to Loop Through a List:
1. **Looping by Item:** `for item in list_name:`
2. **Looping by Index:** `for i in range(len(list_name)):`

Let's test both methods below!

````python
# Array Traversal Examples

print("=== Method 1: Looping by Item ===")
for item in backpack:
    print(f"🎒 Item: {item}")

print("\n=== Method 2: Looping by Index (with Position Labels) ===")
for i in range(len(backpack)):
    print(f"Index {i} -> {backpack[i]}")

print("\n=== Processing Data During Traversal (Calculating Average Score) ===")
total_score = 0
for score in quiz_scores:
    total_score += score

average_score = total_score / len(quiz_scores)
print(f"Total Points: {total_score}")
print(f"Average Score: {average_score:.2f}")
````

````python
# Activity 2: Process Your Data (Input + Computation/Search with Arrays/Lists)

# Do your Activity 2 here
# Empty list (of devices in this case) at the start
five_devices_list = []

# Ask user to enter devices (5 in this case) for the list
print("Enter 5 devices:")
for i in range(5):
    device = input(f"Device {i + 1}: ")
    five_devices_list.append(device)

# Display the user's list of 5 devices
print("\nOriginal list of 5 devices:", five_devices_list)

# Ask user to input index (0-4) of device desired to change/update
device_index = int(input("\nDevice index to update (0-4): "))
new_device = input("New device: ")

# Ask user to input the start and end indexes of desired slice where the user wants to search for an item in the list
slice_start = int(input("Starting Device Index of desired slice where you want to search in: "))
slice_end = int(input("Ending Device Index of desired slice where you want to search in: "))

# Display the desired slice
sliced_devices = five_devices_list[slice_start:slice_end]
print(f"\nSliced Device List ({slice_start}:{slice_end}):", sliced_devices)

# Ask user to input the item in the slice he/she is searching for
search_device = input("\nDevice to search: ")

# Search result feedback that tells whether the item the user is searching for is in the slice the user entered
if search_device in sliced_devices:
    print(f"Device Search Result: {search_device} is found in the sliced device list.")
else:
    print(f"Device Search Result: {search_device} is not found in the sliced device list.")

# Updates the original list by replacing the existing value at the selected index with the new inputted value
five_devices_list[device_index] = new_device

# Display the user's new list of 5 devices
print("\nUpdated list of 5 devices:", five_devices_list)
````

## ✂️ Part 4: List Slicing & Searching

### 1. List Slicing (`list[start:end]`)
Slicing allows you to extract a sub-list (a portion of your list).
- Syntax: `list_name[start:end]`
- **Remember:** The slice starts at `start` index and goes up to, but **does NOT include**, `end` index!
- Example: `backpack[1:4]` gets items at index 1, 2, and 3.

### 2. Linear Searching
Searching means checking if a target value exists in your list and finding its index position.

````python
# Slicing and Searching Demo (Matching Activity 2 Requirements)

# Reset list
backpack = ["Notebook", "Pen", "Tablet", "Laptop", "Book"]

print("Original Backpack List:", backpack)

# Extracting a sliced portion from index 1 to 4 (indexes 1, 2, 3)
sliced_backpack = backpack[1:4]
print("Sliced Sub-List (1:4):", sliced_backpack)

# Linear Search Function
search_item = "Laptop"

if search_item in sliced_backpack:
    found_index = sliced_backpack.index(search_item)
    print(f"🔍 Search Result: '{search_item}' was found in the sliced list at sub-index {found_index}!")
else:
    print(f"🔍 Search Result: '{search_item}' was not found in the sliced list.")
````

## 🧰 Part 5: Essential Python Built-in List Methods

Python provides powerful built-in methods and functions to modify and inspect lists without writing long loops!

| Method / Function | What It Does | Example |
| :--- | :--- | :--- |
| **`len(list)`** | Returns total number of items | `len(backpack)` |
| **`sum(list)`** | Calculates sum of numeric items | `sum(quiz_scores)` |
| **`max(list)` / `min(list)`** | Finds highest / lowest value | `max(quiz_scores)` |
| **`append(item)`** | Adds item to the **end** of list | `backpack.append("Binder")` |
| **`insert(index, item)`** | Inserts item at specific position | `backpack.insert(1, "Ruler")` |
| **`remove(item)`** | Removes first occurrence of item | `backpack.remove("Pen")` |
| **`pop(index)`** | Removes & returns item at index | `backpack.pop(0)` |
| **`sort()` / `sorted()`** | Sorts list in ascending order | `quiz_scores.sort()` |
| **`reverse()`** | Reverses order of list items | `backpack.reverse()` |

## 🏆 Part 7: Mini-Project — Student Score Analyzer Mini-App

Create a complete modular text-based program that manages and analyzes student scores!

This program uses a **modular structure with functions** and a clear **interactive menu** matching Activity 3 (Data Analysis Mini-App).

Run the cell below to test the Student Score Analyzer!

````python
# Activity 3 (Graded or Non-Graded Formative Assessment): Data Analysis Mini App

# Do your Activity 3 here
# Import sys module for an exiting function
import sys

# Introduce the program (a device manager in this case) to the user
print("=== WELCOME TO DEVICE MANAGER! ===")

# Empty array/list of devices at the start
device_list = []

# Functions
# This function adds devices to the array/list of devices in Device Management
def add_devices():
  while True:
    device = input("\nEnter a device to add to Device Management (or type 'done' to finish adding): ").strip()
    if device.lower() == "done":
      print("\nFinished adding devices!")
      break

    elif device == "":
      print("\nInput cannot be empty! Please try again.")

    else:
      device_list.append(device)
  display_devices()
  return_to_main_menu()

# This function displays the main menu of the program and lets the user choose an operation to be executed. The program goes back to the main menu after each execution, which can be ended by exiting (6).
def main_menu():
  print("\n=== DEVICE MANAGEMENT ===")
  print("1. Add Device/s")
  print("2. Display Devices")
  print("3. Count Devices")
  print("4. Search for a Device")
  print("5. Update a Device")
  print("6. Exit")
  while True:
    activity = (input("Enter the number of the action you want to do: "))
    if activity.strip().lower() in ["1", "one"]:
      add_devices()

    elif activity.strip().lower() in ["2", "two"]:
      display_devices()
      while True:
        back_to_main_menu = input("Would you like to return to the main menu? Y/Yes/yes or N/No/no?\n")
        if back_to_main_menu.strip().lower() in ["y", "yes"]:
          return_to_main_menu()

        elif back_to_main_menu.strip().lower() in ["n", "no"]:
          print("Don't worry, no rush!")

        else:
          print("Invalid input! Please try again.")

    elif activity.strip().lower() in ["3", "three"]:
      count_devices()

    elif activity.strip().lower() in ["4", "four"]:
      search_devices()

    elif activity.strip().lower() in ["5", "five"]:
      update_device()

    elif activity.strip().lower() in ["6", "six"]:
      print("Thank you for using Device Manager and goodbye!")
      print("CLOSING DEVICE MANAGER...")
      sys.exit()

    else:
      print("Invalid input! Please try again.")

# This function sends the user back to the main menu
def return_to_main_menu():
  print("RETURNING TO MAIN MENU...")
  main_menu()

# This function displays the array/list of devices
def display_devices():
  print("\n=== DEVICE LIST ===")
  if len(device_list) == 0:
    while True:
      add_or_not = input("No devices found!\nPlease add at least 1 device to Device Management in order to display! Add device/s? Y/Yes/yes or N/No/no?\n")
      if add_or_not.strip().lower() in ["y", "yes"]:
        add_devices()
        break

      elif add_or_not.strip().lower() in ["n", "no"]:
        return_to_main_menu()
        break

      else:
        print("Invalid input! Please try again.")

  else:
    for i in range(len(device_list)):
      print(f"{i + 1}. {device_list[i]}")

# This function counts and displays the count of devices
def count_devices():
  print("\n=== DEVICE COUNT ===")

  if len(device_list) == 0:
    while True:
      add_or_not = input("No devices counted!\nPlease add at least 1 device to Device Management in order to count! Add device/s? Y/Yes/yes or N/No/no?\n")
      if add_or_not.strip().lower() in ["y", "yes"]:
        add_devices()
        break

      elif add_or_not.strip().lower() in ["n", "no"]:
        return_to_main_menu()
        break

      else:
        print("Invalid input! Please try again.")

  else:
    print(f"Total number of devices: {len(device_list)}")
  return_to_main_menu()

# This function lets the user search for something in the array/list of devices
def search_devices():
  print("\n=== DEVICE SEARCH ===")
  if len(device_list) == 0:
    while True:
      add_or_not = input("No device available to search!\nPlease add at least 1 device to Device Management in order to search! Add device/s? Y/Yes/yes or N/No/no?\n")
      if add_or_not.strip().lower() in ["y", "yes"]:
        add_devices()
        break

      elif add_or_not.strip().lower() in ["n", "no"]:
        return_to_main_menu()
        break

      else:
        print("Invalid input! Please try again.")

  else:
    while True:
      search_device = input("Enter device to search (or type 'cancel' to stop searching): ").strip()
      if search_device.lower() == "cancel":
        print("Search cancelled!")
        break

      elif search_device == "":
        print("Input cannot be empty! Please try again.")

      elif search_device.lower() in [device.lower() for device in device_list]:
        print(f"{search_device} was found in Device Manager")

      else:
        print(f"{search_device} was not found! Please try again.")
    return_to_main_menu()

# This function lets the user change or remove a device from the array/list based on the inputted index in the Device Management
def update_device():
  print("\n=== UPDATE DEVICES ===")
  if len(device_list) == 0:
    while True:
      add_or_not = input("No device available to update!\nPlease add at least 1 device to Device Manager in order to be able to update a device! Add device/s? Y/Yes/yes or N/No/no?\n")
      if add_or_not.strip().lower() in ["y", "yes"]:
        add_devices()
        break

      elif add_or_not.strip().lower() in ["n", "no"]:
        return_to_main_menu()
        break

      else:
        print("Invalid input! Please try again.")

  else:
    display_devices()
    update = input("Which form of updating do you want to execute?\n1. Change\n2. Remove\n")
    if update.strip().lower() in ["c", "change", "1"]:
      choice_change = int(input("Enter the number of the device you want to change: "))
      index_change = choice_change - 1
      new_device = input(f"Enter the new device name you want to change device {choice_change} with: ").strip()
      device_list[index_change] = new_device
      print(f"Successfully changed device {choice_change} with {new_device}!")
      display_devices()

    elif update.strip().lower() in ["r", "remove", "2"]:
      choice_remove = int(input("Enter the number of the device you want to remove: "))
      index_remove = choice_remove - 1
      remove_device = device_list.pop(index_remove)
      print(f"Successfully removed {remove_device} from Device Management!")
      display_devices()

    else:
      print("Invalid input! Please try again.")

    return_to_main_menu()

# Starts the program with the main menu
main_menu()
````
